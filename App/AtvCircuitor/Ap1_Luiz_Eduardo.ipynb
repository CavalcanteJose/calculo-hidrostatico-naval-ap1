### Formulário de Apoio
* **Relação Frequência-Período:** $T = \frac{1}{f} \iff f = \frac{1}{T}$
* **Pulsação / Frequência Angular:** $\omega = \frac{2\pi}{T} = 2\pi f$
* **Valor Médio Temporal:** $\langle x(t) \rangle = X_{\text{méd}} = \frac{1}{T} \int_{0}^{T} x(t)\,dt$
* **Valor Eficaz (Root Mean Square):** $X_{\text{rms}} = \sqrt{\frac{1}{T} \int_{0}^{T} x^2(t)\,dt}$
* **Senoide Pura:** $V_{\text{rms}} = \frac{V_m}{\sqrt{2}} \approx 0,7071 V_m$, $V_{\text{méd, meio ciclo}} = \frac{2V_m}{\pi} \approx 0,6366 V_m$
* **Notação Fasorial & Identidades:** $\cos(\omega t + \theta) \leftrightarrow V_m \angle \theta$; $\sin(\omega t + \alpha) = \cos(\omega t + \alpha - 90^\circ)$

### Questão 1
Cálculo analítico do período temporal a partir da frequência fundamental $T = f^{-1}$:
* **Item a:** $f = 200\text{ Hz} \implies T = \frac{1}{200} = 5\times 10^{-3}\text{ s} = 5\text{ ms}$
* **Item b:** $f = 40\text{ MHz} = 4\times 10^7\text{ Hz} \implies T = \frac{1}{4\times 10^7} = 2,5\times 10^{-8}\text{ s} = 25\text{ ns}$
* **Item c:** $f = 20\text{ kHz} = 2\times 10^4\text{ Hz} \implies T = \frac{1}{2\times 10^4} = 5\times 10^{-5}\text{ s} = 50\ \mu\text{s}$
* **Item d:** $f = 1\text{ Hz} \implies T = \frac{1}{1} = 1,0\text{ s}$


```python
# Q1: Implementação orientada a dicionários e formatação com unidades de engenharia
dados_q1 = {
    'a': ('200 Hz', 200.0),
    'b': ('40 MHz', 40.0e6),
    'c': ('20 kHz', 20.0e3),
    'd': ('1 Hz', 1.0)
}

def formatar_tempo(t_seg):
    if t_seg >= 1:
        return f"{t_seg:.4f} s"
    elif t_seg >= 1e-3:
        return f"{t_seg*1e3:.4f} ms"
    elif t_seg >= 1e-6:
        return f"{t_seg*1e6:.4f} us"
    else:
        return f"{t_seg*1e9:.4f} ns"

print("--- Resolução da Questão 1 ---")
for chave, (rotulo, freq) in dados_q1.items():
    periodo = 1.0 / freq
    print(f"[{chave}] f = {rotulo:<10} -> T = {formatar_tempo(periodo)} ({periodo:.2e} s)")
```

### Questão 6
Para $i(t) = 125\cos(800t + 36,87^\circ)\text{ mA}$:
1. **(a) Frequência:** $f = \frac{800}{2\pi} \approx 127,324\text{ Hz}$
2. **(b) Período:** $T = \frac{1}{f} = 7,854\text{ ms}$
3. **(c) Amplitude máxima:** $I_m = 125\text{ mA}$
4. **(d) Corrente inicial:** $i(0) = 125\cos(36,87^\circ) = 125(0,8) = 100,0\text{ mA}$
5. **(e) Fase inicial:** $\phi = 36,87^\circ = 36,87 \times \frac{\pi}{180} = 0,6435\text{ rad}$
6. **(f) Menor $t > 0$ com $i(t) = 0$:**
   $$\cos(800t + \phi) = 0 \implies 800t + \phi = \frac{\pi}{2} \implies t = \frac{\pi/2 - 0,6435}{800} = 1,1591\text{ ms}$$
7. **(g) Menor $t > 0$ com $\frac{di}{dt} = 0$:**
   $$\frac{di}{dt} = -125(800)\sin(800t + \phi) = 0 \implies 800t + \phi = \pi \implies t = \frac{\pi - 0,6435}{800} = 3,1226\text{ ms}$$


```python
# Q6: Resolução algébrica e pontos críticos no gráfico
w6 = 800.0
Im6 = 125.0
phi_deg = 36.87
phi_rad = np.radians(phi_deg)

f6 = w6 / (2 * np.pi)
T6_ms = (1.0 / f6) * 1000
i0 = Im6 * np.cos(phi_rad)
t_raiz_ms = ((np.pi/2 - phi_rad) / w6) * 1000
t_crit_ms = ((np.pi - phi_rad) / w6) * 1000

print("--- Resolução da Questão 6 ---")
print(f"(a) f = {f6:.3f} Hz")
print(f"(b) T = {T6_ms:.3f} ms")
print(f"(c) Im = {Im6:.1f} mA")
print(f"(d) i(0) = {i0:.2f} mA")
print(f"(e) phi = {phi_deg:.2f} graus ({phi_rad:.4f} rad)")
print(f"(f) Menor t > 0 onde i(t)=0: t = {t_raiz_ms:.4f} ms")
print(f"(g) Menor t > 0 onde di/dt=0: t = {t_crit_ms:.4f} ms")

t_plot = np.linspace(0, T6_ms, 1000)
i_plot = Im6 * np.cos(w6 * (t_plot/1000) + phi_rad)

fig, ax = plt.subplots(figsize=(9, 4))
ax.plot(t_plot, i_plot, color='#34495e', lw=2, label=r'$i(t) = 125\cos(800t + 36.87^\circ)\ \mathrm{mA}$')
ax.plot(0, i0, 'o', color='purple', label=f'i(0) = {i0:.1f} mA')
ax.plot(t_raiz_ms, 0, 's', color='crimson', label=f'Raiz: t = {t_raiz_ms:.3f} ms')
ax.plot(t_crit_ms, -Im6, '^', color='teal', label=f'Minimo: t = {t_crit_ms:.3f} ms')
ax.axhline(0, color='gray', lw=0.8, ls=':')
ax.set_title('Questão 6 — Identificação de Pontos Críticos e Condição Inicial')
ax.set_xlabel('Tempo (ms)')
ax.set_ylabel('Corrente i(t) (mA)')
ax.legend(loc='upper right')
plt.tight_layout()
plt.show()
```
