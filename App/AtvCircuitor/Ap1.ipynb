### Fórmulas Fundamentais
* Período e Frequência: $T = \frac{1}{f}$, $f = \frac{1}{T}$
* Frequência Angular: $\omega = \frac{2\pi}{T} = 2\pi f$
* Tensão Pico a Pico: $V_{pp} = 2 V_m$
* Valor Médio (meio ciclo de senoide pura): $V_{\text{méd}} = \frac{2V_m}{\pi} \approx 0,637 V_m$
* Valor Eficaz (RMS para senoide pura): $V_{\text{rms}} = \frac{V_m}{\sqrt{2}} \approx 0,707 V_m$
* Relações trigonométricas: $\sin(\omega t \pm 90^\circ) = \cos(\omega t \mp 90^\circ)$, $\cos(\omega t - 90^\circ) = \sin(\omega t)$

### Questão 2
Se uma forma de onda periódica tem uma frequência de $1\text{ kHz}$, qual o tempo (em segundos) necessário para completar 5 ciclos?

**Passo a passo:**
1. Frequência $f = 1\text{ kHz} = 1000\text{ Hz}$.
2. Período de 1 ciclo: $T = \frac{1}{f} = \frac{1}{1000} = 0,001\text{ s} = 1\text{ ms}$.
3. Tempo total para 5 ciclos: $\Delta t = 5 \cdot T = 5 \cdot 0,001 = 0,005\text{ s} = 5\text{ ms}$.


```python
# Questão 2: Tempo para 5 ciclos
f = 1000.0  # 1 kHz
T = 1.0 / f
t_5_ciclos = 5 * T

print("=== Questão 2 ===")
print(f"Frequência: {f} Hz")
print(f"Período de 1 ciclo: {T} s ({T*1000:.1f} ms)")
print(f"Tempo para completar 5 ciclos: {t_5_ciclos} s ({t_5_ciclos*1000:.1f} ms)")

# Simulação gráfica de 5 ciclos
t = np.linspace(0, t_5_ciclos, 1000)
v = np.sin(2 * np.pi * f * t)
plt.figure()
plt.plot(t * 1000, v, 'b-', label='Sinal f = 1 kHz')
plt.axvline(x=t_5_ciclos*1000, color='r', linestyle='--', label=f'5 ciclos = {t_5_ciclos*1000:.1f} ms')
for k in range(1, 6):
    plt.axvline(x=k*T*1000, color='gray', linestyle=':', alpha=0.5)
plt.title('Questão 2: 5 ciclos de onda senoidal de 1 kHz')
plt.xlabel('Tempo (ms)')
plt.ylabel('Amplitude')
plt.legend()
plt.show()
```

### Questão 6
Uma corrente senoidal é dada pela expressão $i(t) = 125\cos(800t + 36,87^\circ)\text{ mA}$.
Determine:
* **(a)** $f$ em hertz: $f = \frac{800}{2\pi} \approx 127,32\text{ Hz}$
* **(b)** $T$ em milissegundos: $T = \frac{1}{f} = \frac{2\pi}{800} \approx 7,854\text{ ms}$
* **(c)** $I_m$: $125\text{ mA}$
* **(d)** $i(0)$: $125\cos(36,87^\circ) = 125 \cdot 0,8 = 100\text{ mA}$
* **(e)** $\phi$ em graus e radianos: $\phi = 36,87^\circ = 36,87 \cdot \frac{\pi}{180} \approx 0,6435\text{ rad}$
* **(f)** o menor valor positivo de $t$ em que $i = 0$:  
  $800t + 36,87^\circ = 90^\circ \implies 800t = 53,13^\circ = 0,9273\text{ rad} \implies t = \frac{0,9273}{800} \approx 1,159\text{ ms}$
* **(g)** o menor valor positivo de $t$ em que $di/dt = 0$:  
  $\frac{di}{dt} = -125 \cdot 800 \sin(800t + 36,87^\circ) = 0 \implies 800t + 36,87^\circ = 180^\circ \implies 800t = 143,13^\circ = 2,498\text{ rad} \implies t \approx 3,123\text{ ms}$


```python
# Questão 6: Análise completa de i(t) = 125 cos(800t + 36.87°) mA
Im = 125.0
w = 800.0
phi_deg = 36.87
phi_rad = np.radians(phi_deg)

f = w / (2 * np.pi)
T_ms = (1.0 / f) * 1000
i_0 = Im * np.cos(phi_rad)

# (f) i = 0 => 800t + phi = pi/2
t_zero_ms = ((np.pi/2 - phi_rad) / w) * 1000

# (g) di/dt = 0 => 800t + phi = pi (primeiro pico negativo após t=0)
t_extremo_ms = ((np.pi - phi_rad) / w) * 1000

print("=== Questão 6: Resultados ===")
print(f"(a) Frequência (f): {f:.2f} Hz")
print(f"(b) Período (T): {T_ms:.3f} ms")
print(f"(c) Amplitude máxima (Im): {Im:.1f} mA")
print(f"(d) i(0): {i_0:.2f} mA")
print(f"(e) Fase (phi): {phi_deg:.2f} graus = {phi_rad:.4f} rad")
print(f"(f) Menor t > 0 para i(t) = 0: {t_zero_ms:.4f} ms")
print(f"(g) Menor t > 0 para di/dt = 0: {t_extremo_ms:.4f} ms")

# Gráfico
t_axis = np.linspace(0, T_ms, 1000)
i_vals = Im * np.cos(w * (t_axis/1000) + phi_rad)

plt.figure()
plt.plot(t_axis, i_vals, 'b-', lw=2, label=r'$i(t) = 125\cos(800t + 36.87^\circ)$ mA')
plt.plot(0, i_0, 'ro', label=f'i(0) = {i_0:.1f} mA')
plt.plot(t_zero_ms, 0, 'go', label=f'i(t)=0 em {t_zero_ms:.3f} ms')
plt.plot(t_extremo_ms, -Im, 'mo', label=f'di/dt=0 (mínimo) em {t_extremo_ms:.3f} ms')
plt.axhline(0, color='k', lw=0.5)
plt.title('Questão 6: Forma de Onda e Pontos Notáveis')
plt.xlabel('Tempo (ms)')
plt.ylabel('Corrente (mA)')
plt.legend()
plt.show()
```

### Questão 12
Expresse $v(t) = \cos 5t \cdot \sin(3t + 45^\circ)$ como uma soma de duas funções cosseno e encontre o seu período.

**Passo a passo:**
1. Usando a identidade $\sin A \cos B = \frac{1}{2}[\sin(A+B) + \sin(A-B)]$:
   $$v(t) = \frac{1}{2}[\sin(8t + 45^\circ) + \sin(-2t + 45^\circ)] = \frac{1}{2}\sin(8t + 45^\circ) - \frac{1}{2}\sin(2t - 45^\circ)$$
2. Convertendo cada termo para cosseno:
   * $\frac{1}{2}\sin(8t + 45^\circ) = \frac{1}{2}\cos(8t + 45^\circ - 90^\circ) = 0,5\cos(8t - 45^\circ)$
   * $-\frac{1}{2}\sin(2t - 45^\circ) = \frac{1}{2}\cos(2t - 45^\circ + 90^\circ) = 0,5\cos(2t + 45^\circ)$
   $$v(t) = 0,5\cos(8t - 45^\circ) + 0,5\cos(2t + 45^\circ)$$
3. Frequências: $\omega_1 = 8\text{ rad/s}$ e $\omega_2 = 2\text{ rad/s}$.
   $$\omega_0 = \text{MDC}(8, 2) = 2\text{ rad/s} \implies T_0 = \frac{2\pi}{\omega_0} = \pi \approx 3,1416\text{ s}$$


```python
# Questão 12: Verificação da decomposição e período
t = np.linspace(0, 2*np.pi, 1000)
v_orig = np.cos(5*t) * np.sin(3*t + np.radians(45))
v_decomp = 0.5 * np.cos(8*t - np.radians(45)) + 0.5 * np.cos(2*t + np.radians(45))

print("=== Questão 12 ===")
print("Erro máximo entre produto e soma de cossenos:", np.max(np.abs(v_orig - v_decomp)))
print(f"Período fundamental T0 = pi ~= {np.pi:.4f} s")

plt.figure()
plt.plot(t, v_orig, 'b-', lw=2, label='v(t) original')
plt.plot(t, v_decomp, 'r--', lw=2, label='0.5 cos(8t - 45°) + 0.5 cos(2t + 45°)')
plt.axvline(np.pi, color='g', linestyle=':', label='T = pi ~= 3.14 s')
plt.title('Questão 12: Decomposição e Período de $v(t)$')
plt.xlabel('Tempo (s)')
plt.ylabel('Amplitude')
plt.legend()
plt.show()
```

### Questão 28
Determine o valor eficaz da corrente periódica da Figura: sobe linearmente de $0$ a $20\text{ A}$ entre $t = 0$ e $t = 80\text{ ms}$; desce linearmente de $20\text{ A}$ a $0$ entre $t = 80\text{ ms}$ e $t = 100\text{ ms}$; repete-se com período $T = 100\text{ ms}$.

**Passo a passo:**
* Como cada parte é uma rampa linear entre $0$ e o pico $I_m$:
  $$\int_0^{T_1} \left(\frac{I_m t}{T_1}\right)^2 dt = \frac{I_m^2 T_1}{3},\quad \int_0^{T_2} (\text{rampa})^2 dt = \frac{I_m^2 T_2}{3}$$
* Integral em todo o período $T = T_1 + T_2$:
  $$\int_0^T i^2(t)\,dt = \frac{I_m^2(T_1 + T_2)}{3} = \frac{I_m^2 T}{3}$$
$$I_{\text{rms}} = \sqrt{\frac{I_m^2}{3}} = \frac{I_m}{\sqrt{3}} = \frac{20}{\sqrt{3}} \approx 11,547\text{ A}$$


```python
# Questão 28: Onda dente-de-serra assimétrica
Im = 20.0
T_ms = 100.0
t1_ms = 80.0

I_rms_28 = Im / np.sqrt(3)

print("=== Questão 28 ===")
print(f"Corrente Máxima Im = {Im} A")
print(f"Corrente Eficaz I_rms = 20/sqrt(3) ~= {I_rms_28:.4f} A")

def i_onda_q28(t_ms):
    t_mod = t_ms % 100
    if t_mod <= 80:
        return (20.0 / 80.0) * t_mod
    else:
        return 20.0 - (20.0 / 20.0) * (t_mod - 80)

t_arr = np.linspace(0, 200, 1000)
i_arr = np.array([i_onda_q28(t) for t in t_arr])

plt.figure()
plt.plot(t_arr, i_arr, 'b-', lw=2, label='i(t) [A]')
plt.axhline(I_rms_28, color='r', linestyle='--', label=f'I_rms = {I_rms_28:.2f} A')
plt.title('Questão 28: Corrente com Subida em 80 ms e Descida em 20 ms')
plt.xlabel('Tempo (ms)')
plt.ylabel('Corrente (A)')
plt.legend()
plt.show()
```

Calcule o valor eficaz da tensão dada pela forma de onda mostrada na Figura:

**Passo a passo:**
$$V_{\text{rms}} = \sqrt{\frac{1}{T}\int_0^T v^2(t)\,dt} = \sqrt{\frac{1}{5}\int_0^1 10^2\,dt} = \sqrt{\frac{100 \times 1}{5}} = \sqrt{20} = 2\sqrt{5} \approx 4,4721\text{ V}$$
